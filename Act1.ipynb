# Análisis exploratorio sobre obesidad en países latinoamericanos.

### Introducción

Este reporte analiza los diferentes datos sobre obesidad en 3 países de latinoamérica diferentes: Perú, México y Colombia. Los datos provienen de una recopilación creada en la Universidad de la Costa de Colombia. Los datos fueron recopilados debido a una creciente preocupación sobre la obesidad y en este reporte se estarán analizando e interpretando de diferentes maneras donde se llegue a una conclusión de como crece o decrece esta enfermedad y características que influyen en tener obesidad.


### Exploración inical

*   Sexo
*   Edad
*   Estatura
*   Peso
*   Familiar con sobrepeso
*   Si la persona come muchas calorías
*   Si come vegetales

Los datos son características personales de cada persona que varían segun su genética pero también sus habitos, y se provee el nivel de obesidad al lado para saber como se relacionan entre si todas sus características y su estado de forma actual.
Todos estos datos son recopilados para establecer relaciones importantes en las características que influyen en que una persona padezca obesidad y nos permiten establecer diversas correlaciones.



###Reflexión sobre el diseño del estudio
En esta investigación, como parte del equipo de investigación, sería relevante añadir la variable de pasos diarios de cada una de las personas. Esta variable permitiría saber sobre cuanto movimiento tiene esa persona, las calorías que quema andando,y su tipo de vida, si es más o menos sedentario. Con esta variable se puede inferir la relación entre actividad física y obesidad debido a los datos extra que podemos obtener, por ejemplo si una persona camina 10000 pasos diarios, se puede calcular las calorias quemadas, su estilo de vida, y esos son datos que pueden ayudar a inferir la obesidad de una persona.


###Definición de una variable de interés para inferencia
Como líder del proyecto, considero que la variable de Peso como variable de salida del estudio. Esto debido a que es una variable cuantitativa que representa el dato físico más preciso para medir el estado de una persona. Esta variable permite recolectar datos reales que facilitan la inferencia mediante el entendimiento de qué valores influyen en la masa corporal de un individuo y el riesgo de obtener obesidad.

#### Análisis descriptivo básico
Para realizar un análisis descriptivo, se utilizaría una gráfica que relacione la estatura con el peso mediante guna gráfica de dispersión. Esto permite observar una correlación directa entre como varía el peso y la altura de las personas. La agrupación de puntos permitirá observar la concentración en el rango númerico.

###Análisis gráfico enfocado en una subpoblación

```python
import matplotlib.pyplot as plt

# Filter the DataFrame for 'Female' gender
df_female = df[df['Sexo'] == 'Female']

plt.figure(figsize=(10, 6))
plt.scatter(df_female['Estatura'], df_female['Peso'], alpha=0.7)
plt.title('Gráfico de Dispersión: Estatura vs. Peso (Solo Mujeres)')
plt.xlabel('Estatura (m)')
plt.ylabel('Peso (kg)')
plt.grid(True)
plt.show()
```

Para este análisis se seleccionó a la población femenina del conjunto de datos con el fin de explorar la relación entre la estatura y el peso, definida como la variable de respuesta. A través de una gráfica de dispersión, se observa una tendencia ascendente que indica una correlación positiva donde, generalmente, a mayor estatura corresponde un mayor peso corporal. La mayoría de las observaciones se concentran en un rango de estatura de entre 1.55 m y 1.65 m, destacando una dispersión significativa en este intervalo. Dicha variabilidad muestra que individuos con la misma altura presentan pesos que oscilan desde los 50 kg hasta más de 120 kg, lo que sugiere que la estatura no es el único factor determinante en la masa corporal. Asimismo, se identifican patrones de datos extremos con puntos aislados en niveles superiores de peso que representan casos de riesgo de obesidad elevado, permitiendo inferir que existen otras variables externas que influyen directamente en el estado físico de la subpoblación analizada.
